```python
import os
import sys

import asyncio
from dotenv import load_dotenv

from smolagents import GradioUI

from athos import build_config, build_flow
```

```python
# Pickup conda environment executables
bin_dir = os.path.join(sys.prefix, "bin")
os.environ["PATH"] = bin_dir + os.pathsep + os.environ["PATH"]
```

```python
# Load some settings from a .env file
load_dotenv('../.env')
```

```python
# Setup CBorg
api_base="https://api.cborg.lbl.gov"
api_key_path = '~/.cborg/api_key'

with open(os.path.expanduser(api_key_path)) as f:
    api_key = f.read().strip()

os.environ['CBORG_API_URL'] = api_base
os.environ['CBORG_API_KEY'] = api_key
```

```python
# Load ATHOS configuration
config = build_config('../config/config.yaml')
```

```python
from athos.main import _format_config
print(_format_config(config))
```

```python
docs_agent_flow = build_flow('smol_docs_agent', config)
```

```python
port = 8989  # Or any available port
root_path = f"{os.environ['JUPYTERHUB_SERVICE_PREFIX']}proxy/{port}/"
proxy_url = f"https://jupyter.nersc.gov{root_path}"
```

```python
print(f"Access your Gradio app at: {proxy_url}")
GradioUI(docs_agent_flow.agent).launch(
    server_name="0.0.0.0",
    server_port=port,
    share=False,
    root_path=root_path,
    inline=False
)
```
